```python
# 作业1：热身：定义简单函数
def say_hi():
    print("Hello, world!")

def goat_pad(s):
    print("goat" + s + "goat")

def print_n(s, n):
    for _ in range(n):
        print(s)

print("--- 测试热身函数 ---")
say_hi()
goat_pad("hello")
print_n("Python", 3)


# 作业1：欧几里得算法
def gcd(m, n):
    # 取绝对值，防止负数导致死循环
    m = abs(m)
    n = abs(n)
    while n != 0:
        m, n = n, m % n
    return m

print("\n--- 测试 GCD 算法 ---")
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300) =", gcd(1200, 300))
print("gcd(5040, 60) =", gcd(5040, 60))
```

```python
import math

# 作业1：近似欧拉数 e
def euler_limit(n):
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1 / fact
    return total

def euler_approx(epsilon):
    total = 1.0
    fact = 1
    k = 1
    # 只要误差大于等于 epsilon，就继续累加
    while abs(total - math.exp(1)) >= epsilon:
        fact *= k
        total += 1 / fact
        k += 1
    return total

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"前 {k} 项和: {euler_infinite_sum(k)}")

print("--- 测试近似欧拉数 ---")
print("euler_limit(100) =", euler_limit(100))
print("euler_infinite_sum(4) =", euler_infinite_sum(4))
print("euler_approx(0.001) =", euler_approx(0.001))
print("\n--- 打印 euler_sum_table ---")
print_euler_sum_table(5)

# 作业1：测试整数属性
def is_power_of_2(n):
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1

def is_power(b, n):
    if b < 1 or n < 1:
        return False
    if b == 1:
        return n == 1
    while n % b == 0:
        n //= b
    return n == 1

print("\n--- 测试整数属性 ---")
print("is_power_of_2(8) =", is_power_of_2(8))
print("is_power_of_2(10) =", is_power_of_2(10))
print("is_power(3, 27) =", is_power(3, 27))
print("is_power(3, 10) =", is_power(3, 10))
```
