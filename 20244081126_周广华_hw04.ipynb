# 第一部分：正则表达式练习

```python
import re

# 1. 检查字符串是否仅包含英文字母和数字
def check_alnum(s):
    return bool(re.fullmatch(r'[a-zA-Z0-9]+', s))

print("1. 检查字母和数字：")
print(check_alnum("Hello123"))  # True
print(check_alnum("Hello_123")) # False
```

```python
# 2. 匹配一个 a 后面跟着零个或多个 b 的字符串
def match_ab_star(s):
    return bool(re.fullmatch(r'ab*', s))

print("\n2. a 后面跟零个或多个 b：")
print(match_ab_star("a"), match_ab_star("abbb"), match_ab_star("ac")) # True True False
```

```python
# 3. 匹配一个 a 后面跟着一个或多个 b 的字符串
def match_ab_plus(s):
    return bool(re.fullmatch(r'ab+', s))

print("\n3. a 后面跟一个或多个 b：")
print(match_ab_plus("a"), match_ab_plus("abbb"), match_ab_plus("ac")) # False True False
```

```python
# 4. 匹配以 a 开头、以 b 结尾，中间任意字符的字符串
def match_a_any_b(s):
    return bool(re.fullmatch(r'a.*b', s))

print("\n4. a 开头 b 结尾：")
print(match_a_any_b("ab"), match_a_any_b("a123b"), match_a_any_b("abc")) # True True False
```

```python
# 5. 删除 IP 地址中各段数字的前导零
def strip_ip_leading_zeros(ip):
    # 匹配段首的 0，后面跟着数字，替换为后面的数字（例如 001 -> 1，010 -> 10）
    return re.sub(r'\b0+(\d)', r'\1', ip)

print("\n5. 删除 IP 前导零：")
print(strip_ip_leading_zeros("192.168.001.010")) # 192.168.1.10
print(strip_ip_leading_zeros("192.168.000.010")) # 192.168.0.10
```

```python
# 6. 将字符串中的空格替换为下划线，并将下划线替换为空格
def swap_space_underscore(s):
    # 使用一个正则表达式和一个 lambda 函数，避免两步替换导致冲突
    return re.sub(r'(\s)|(_)', lambda m: '_' if m.group(1) else ' ', s)

print("\n6. 空格和下划线互换：")
print(swap_space_underscore("Hello World_Test")) # Hello_World Test
```

```python
# 7. 将 yyyy-mm-dd 格式的日期转换为 dd-mm-yyyy
def convert_date_format(date_str):
    return re.sub(r'(\d{4})-(\d{2})-(\d{2})', r'\3-\2-\1', date_str)

print("\n7. 日期格式转换：")
print(convert_date_format("2024-05-01")) # 01-05-2024
```

```python
# 8. 找出所有以 a 或 e 开头的单词（忽略大小写）
def find_ae_words(s):
    return re.findall(r'\b[aeAE]\w*', s)

print("\n8. 找出 a 或 e 开头的单词：")
print(find_ae_words("An apple a day keeps everyone engaged.")) # ['An', 'apple', 'a', 'everyone', 'engaged']
```

```python
# 9. 将给定字符串中的 Road 缩写为 Rd
def abbreviate_road(s):
    return re.sub(r'\bRoad\b', 'Rd', s)

print("\n9. Road 缩写为 Rd：")
print(abbreviate_road("Main Road, Oak Road, Broadway")) # Main Rd, Oak Rd, Broadway
```

```python
# 10. 找出字符串中所有长度至少为 4 个字符的单词
def find_long_words(s):
    return re.findall(r'\b\w{4,}\b', s)

print("\n10. 找长度至少为 4 的单词：")
print(find_long_words("This is a simple test sentence")) # ['This', 'simple', 'test', 'sentence']
```

```python
# 11. 删除字符串中除字母和数字之外的所有字符
def remove_special_chars(s):
    return re.sub(r'[^a-zA-Z0-9]', '', s)

print("\n11. 删除非字母数字字符：")
print(remove_special_chars("Hello, World! 123.")) # HelloWorld123
```

```python
# 12. 删除字符串中的括号及括号内的内容
def remove_parentheses(s):
    return re.sub(r'\s*\([^)]*\)', '', s)

print("\n12. 删除括号及内容：")
text_12 = """example (.com)
w3resource
github (.com)
stackoverflow (.com)"""
print(remove_parentheses(text_12))
# 预期：example, w3resource, github, stackoverflow
```

# 第二部分：网页检索气候数据

**实际使用的 URL**: `https://en.wikipedia.org/wiki/Madison,_Wisconsin`
**访问日期**: 2026-11-24
**测试页面 URL**: `https://en.wikipedia.org/wiki/Hyderabad`

```python
import requests
from bs4 import BeautifulSoup
from datetime import date
import re

url = "https://en.wikipedia.org/wiki/Madison,_Wisconsin"
print("访问日期:", date.today())

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
resp = requests.get(url, headers=headers)
soup = BeautifulSoup(resp.text, "html.parser")

# 查找表格并打印结构
climate_tables = [t for t in soup.find_all("table") if "Climate data for" in t.get_text()]
print(f"在当前页面找到了 {len(climate_tables)} 个包含气候数据的表格。")

if climate_tables:
    target_table = climate_tables[0]
    children = [child.name for child in target_table.children if child.name is not None]
    print(f"第一个表格的直接子节点数量: {len(children)}")
    print(f"直接子节点的标签名: {children}")
```

```python
def retrieve_climate_table(url):
    """
    接收 URL，若存在气候数据表格返回 Tag 对象，否则返回 None。
    """
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
    # 网络错误或 HTTP 错误会引发异常
    resp = requests.get(url, headers=headers)
    resp.raise_for_status() 
    
    soup = BeautifulSoup(resp.text, "html.parser")
    for table in soup.find_all("table"):
        if "Climate data for" in table.get_text():
            return table
    return None

print("\n测试 retrieve_climate_table:")
# 测试维基百科的其他城市
url_hyd = "https://en.wikipedia.org/wiki/Hyderabad"
table_test = retrieve_climate_table(url_hyd)
print("成功抓取 Hyderabad 的表格" if table_test else "未找到表格")
```

```python
def list_climate_table_row_names(url):
    """
    接收城市页面 URL，返回表格行名组成的列表。找不到表格返回 None。
    """
    table = retrieve_climate_table(url)
    if table is None:
        return None
        
    row_names = []
    for tr in table.find_all("tr"):
        th = tr.find("th")
        if th:
            # 使用正则合并空白字符并去除首尾空格
            text = re.sub(r'\s+', ' ', th.get_text()).strip()
            # 过滤掉表头标题、月份行和数据来源
            if "Climate data for" in text or "Month" in text or "Source" in text:
                continue
            # 如果没有数字（比如纯月份文本或空内容），通常不是我们要找的指标行名
            if text and not text.isdigit():
                row_names.append(text)
    return row_names

print("\n测试 list_climate_table_row_names:")
names = list_climate_table_row_names(url)
print(f"提取到 {len(names)} 个行名：")
print(names) # 应该能看到类似 ['Record high °F (°C)', 'Mean daily maximum °F (°C)', ...] 的列表
```
