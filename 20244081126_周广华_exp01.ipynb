```python
# 1. 成绩等级判断
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在 0~100 之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print("--- 测试成绩等级 ---")
print("95分:", score_level(95))
print("85分:", score_level(85))
print("75分:", score_level(75))
print("55分:", score_level(55))
try:
    score_level(105)
except ValueError as e:
    print("越界测试:", e)

# 2. for 和 while 循环求和
print("\n--- 1到100求和 ---")
total_for = 0
for i in range(1, 101):
    total_for += i
print("for 循环结果:", total_for)

total_while = 0
i = 1
while i <= 101 - 1:
    total_while += i
    i += 1
print("while 循环结果:", total_while)

# 3. 递归与迭代阶乘
print("\n--- 阶乘计算 ---")
def fact_rec(n):
    return 1 if n == 0 else n * fact_rec(n - 1)

def fact_iter(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    print(f"{n}! 递归={fact_rec(n)}, 迭代={fact_iter(n)}")

# 4. summary 函数返回元组
print("\n--- summary 统计 ---")
def summary(values):
    return min(values), max(values), sum(values) / len(values)

low, high, avg = summary([85, 58, 92])
print(f"最低: {low}, 最高: {high}, 平均: {avg:.2f}")

# 5. 可变参数 *args
print("\n--- 可变参数 ---")
def sum_all(*args):
    return sum(args)

print("无参数:", sum_all())
print("单个参数:", sum_all(5))
print("多个参数:", sum_all(1, 2, 3, 4))
```

```python
# 初始模拟数据
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 1. zip() 构建字典并遍历
print("--- 1. 字典遍历 ---")
score_dict = dict(zip(names, scores))
for name, score in score_dict.items():
    print(f"{name}: {score}")

# 2. enumerate() 输出序号
print("\n--- 2. enumerate 遍历 ---")
for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{index}. {name}: {score}")

# 3. 列表推导式
print("\n--- 3. 列表推导式 ---")
passed_names = [name for name, score in zip(names, scores) if score >= 60]
score_squares = [score ** 2 for score in scores]
print("及格学生:", passed_names)
print("成绩平方:", score_squares)

# 4. 基础统计与格式化
print("\n--- 4. 基础统计 ---")
n = len(scores)
avg = sum(scores) / n
pass_rate = sum(1 for s in scores if s >= 60) / n
# 使用 f-string 格式化
print(f"人数: {n}")
print(f"平均分: {avg:.2f}")
print(f"最高分: {max(scores)}, 最低分: {min(scores)}")
print(f"及格率: {pass_rate:.1%}")

# 5. 比较 alias 与 copied
print("\n--- 5. 浅拷贝与直接赋值 ---")
demo = [85, 58, 92]  # 用单独的列表演示
alias = demo
copied = demo.copy()
alias[0] = 0  # 修改 alias
print("原始列表 demo:", demo)   # 发生了改变，因为 alias 指向同一个内存地址
print("副本 copied:", copied)   # 没有改变，因为是独立的内存对象

# 6. 元组与 zip
print("\n--- 6. 元组与 zip ---")
record = ("Alice", 85)
try:
    record[0] = "Bob"
except TypeError as e:
    print("元组不可变:", e)

# 长度不同的序列使用 zip
print("长短序列 zip 结果:", list(zip([1, 2, 3], ["a", "b"])))  # 截断到最短的长度
```

```python
import os
import pickle
import student_utils

# 1. 目录操作
print("--- 1. 目录操作 ---")
print("当前工作目录:", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
data_path = os.path.join(data_dir, "students.txt")

# 2. 写入数据并读取解析
print("\n--- 2. 写入与读取文件 ---")
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(data_path, "w", encoding="utf-8") as f:
    f.write(content)

valid_records = []
invalid_records = []

# 3 & 4. 逐行处理，捕获异常
print("\n--- 3 & 4. 异常处理与解析 ---")
with open(data_path, encoding="utf-8") as f:
    next(f)  # 跳过表头
    for line_no, line in enumerate(f, start=2):
        line = line.strip()
        if not line:
            continue
        try:
            # 使用我们自己写的模块解析
            record = student_utils.parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            invalid_records.append((line_no, line, str(e)))
            
print(f"有效记录数: {len(valid_records)}")
print(f"无效记录: {invalid_records}")

# 5. 写入 report.txt，演示 w 与 a
print("\n--- 5. 写入统计报告 ---")
report_path = os.path.join(data_dir, "report.txt")
valid_scores = [r["score"] for r in valid_records]
avg_score = sum(valid_scores) / len(valid_scores)
pass_rate = sum(1 for s in valid_scores if s >= 60) / len(valid_scores)

# 使用 w 模式（覆盖）
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {len(valid_records)}\n")
    f.write(f"无效记录数: {len(invalid_records)}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rate:.1%}\n")

# 使用 a 模式（追加）
with open(report_path, "a", encoding="utf-8") as f:
    f.write("--- 追加测试 ---\n")

# 6. 使用 pickle 序列化与反序列化
print("\n--- 6. pickle 序列化 ---")
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)

print("读取数据与原数据相等:", loaded_records == valid_records)

# 7. 模块测试（用真实的 student_utils 测试）
print("\n--- 7. 模块测试 ---")
print("有效输入:", student_utils.parse_student("Alice,85"))
try:
    student_utils.parse_student("Grace,abc")
except ValueError as e:
    print("无效输入:", e)

# 8. 捕获 FileNotFoundError
print("\n--- 8. 读取不存在文件 ---")
try:
    with open("不存在的文件.txt", encoding="utf-8") as f:
        pass
except FileNotFoundError:
    print("捕获到异常：文件不存在！")
```

```python
import copy

# 1. 定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


# 2. 为实验4中的每条有效记录创建对象并输出
print("--- 2. 实例化 Student 对象 ---")
student_objects = []
for record in valid_records:
    # 从实验4的字典中提取数据
    stu = Student(record["name"], record["score"])
    student_objects.append(stu)
    status = "及格" if stu.is_passed() else "不及格"
    print(f"{stu} -> {status}")


# 3. 定义 GraduateStudent 子类
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)  # 调用父类初始化
        self.research_topic = research_topic

    def __str__(self):
        # 重写 __str__ 方法
        return f"GraduateStudent(name={self.name}, score={self.score}, topic={self.research_topic})"

print("\n--- 3. 测试子类 ---")
grad_stu = GraduateStudent("Alice", 85, "人工智能")
print(grad_stu)
print("是否及格:", grad_stu.is_passed())


# 4. 对象复制的比较（深浅拷贝）
print("\n--- 4. 对象复制比较 ---")

# 每次比较都从新建的原始对象开始
def test_copy():
    original = {"name": "Alice", "scores": [85, 90]}
    
    # 直接赋值（引用同一对象）
    assigned = original
    
    # 浅拷贝（外层独立，内层共享）
    shallow = copy.copy(original)
    
    # 深拷贝（完全独立）
    deep = copy.deepcopy(original)
    
    # 修改原始对象内部的列表
    assigned["scores"].append(100)
    
    print("修改后 original:", original)
    print("修改后 assigned:", assigned)
    print("浅拷贝 shallow:", shallow)
    print("深拷贝 deep  :", deep)

test_copy()
```
