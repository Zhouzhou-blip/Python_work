```python
# 1. 字符串的乐趣
def is_palindrome(s):
    s = s.replace(" ", "").lower()
    return s == s[::-1]

def is_abecedarian(s):
    s = s.replace(" ", "").lower()
    for i in range(len(s) - 1):
        if s[i] > s[i + 1]:
            return False
    return True

def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return "".join([char for char in s if char not in vowels])

print("is_palindrome('taco cat'):", is_palindrome("taco cat"))
print("is_abecedarian('abcd efgh xyz'):", is_abecedarian("abcd efgh xyz"))
print("remove_vowels('audio'):", remove_vowels("audio"))


# 2. 列表的乐趣
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1] 

def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

print("\nlist_reverse([1, 2, 3]):", list_reverse([1, 2, 3]))
print("is_sorted([1, 2, 3]):", is_sorted([1, 2, 3]))


# 3. 作为向量的元组
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple) or not isinstance(s, (int, float)):
        raise TypeError("第一个参数必须是元组，第二个参数必须是数字")
    return tuple(x * s for x in t)

def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(x) != len(y):
        raise ValueError("向量长度必须一致")
    return float(sum(a * b for a, b in zip(x, y)))

def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    row_len = -1
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if row_len == -1:
            row_len = len(row)
        elif len(row) != row_len:
            return False
        for item in row:
            if not isinstance(item, (int, float)):
                return False
    return True

def mx_vec_mult(mx, v):
    if not check_valid_mx(mx) or not isinstance(v, tuple):
        raise ValueError("矩阵或向量格式不正确")
    if len(mx) > 0 and len(mx[0]) != len(v):
        raise ValueError("矩阵列数与向量长度必须一致")
    return tuple(vec_inner_product(row, v) for row in mx)

print("\nvec_scalar_mult((1, 2), 3):", vec_scalar_mult((1, 2), 3))
print("vec_inner_product((1, 2), (3, 4)):", vec_inner_product((1, 2), (3, 4)))
my_mx = ((1, 2), (3, 4))
print("mx_vec_mult:", mx_vec_mult(my_mx, (1, 1)))


# 4. 更多向量的乐趣（面向对象）
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数")
        if elements is None:
            elements = [0] * dim
        if len(elements) != dim:
            raise ValueError("维度与元素个数不一致")
        self.dim = dim
        self.elements = tuple(elements)

    def __eq__(self, other):
        if not isinstance(other, Vector): return False
        return self.elements == other.elements

    def __lt__(self, other):
        if self.dim != other.dim: raise ValueError("维度不同")
        return self.elements < other.elements

    def __le__(self, other):
        if self.dim != other.dim: raise ValueError("维度不同")
        return self.elements <= other.elements

    def __gt__(self, other):
        if self.dim != other.dim: raise ValueError("维度不同")
        return self.elements > other.elements

    def dot(self, other):
        if not isinstance(other, Vector): raise TypeError("必须与 Vector 对象点乘")
        if self.dim != other.dim: raise ValueError("维度不一致")
        return sum(a * b for a, b in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, [x * other for x in self.elements])
        elif isinstance(other, Vector):
            if self.dim != other.dim: raise ValueError("维度不一致")
            return Vector(self.dim, [a * b for a, b in zip(self.elements, other.elements)])
        else:
            raise TypeError("不支持的类型")

    def __rmul__(self, other):
        return self.__mul__(other)

    def __str__(self):
        return f"Vector({self.dim}, {self.elements})"

print("\n--- Vector 类测试 ---")
v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))
print("v1 * 2:", v1 * 2)
print("2 * v1:", 2 * v1)
print("v1 * v2:", v1 * v2)
print("v1.dot(v2):", v1.dot(v2))
```

维度和元素应该是实例属性，因为每个具体向量的维度和坐标都是独立的，不应该被所有向量共享。
